### 1. Linear Regression using Normal Form

## What is the Normal Equation?

The **Normal Equation** is a closed-form (analytical) solution to linear regression. Instead of iterating, it computes the optimal weights **w** directly in a single step.

### The Formula

Given:
- **X**: feature matrix of shape *(m × n)* — m samples, n features
- **y**: target vector of shape *(m,)*

$$w = (X^T X)^{-1} X^T y$$

### When to use it

| Pros | Cons |
|------|------|
| No learning rate to tune | Matrix inversion is O(n³) — slow for large n |
| Exact solution in one step | Fails if $X^T X$ is singular (non-invertible) |
| No convergence issues | Not suitable for very high-dimensional data |

> **Note:** X must include a column of ones if an intercept (bias) term is needed.

```python
class LRNormal:
    def __init__(self):
        self.theta=None

    def fit(self,X,y):
        X=np.array(X)
        y=np.array(y)

        self.theta=np.linalg.inv(X.T @ X) @ X.T @ y

        return self

    def predict(self,X):
        X=np.array(X)
        return round(X @ self.theta,4)
```

### 2. Linear Regression using Gradient Descent

# using 2 loops

# J(w,b)=(1/m) * (Xw-y)*X

# w=w-alpha*gradient

```python
class LinearRegressionGD:
    def __init__(self,alpha=0.01,iterations=100):
        self.alpha=alpha,
        self.iterations=iterations,
        self.theta=None

    def fit(self,X,y):
        m,n=X.shape
        y=y.reshape(-1,1)
        self.theta=np.zeros((n,1))

        for _ in range(self.iterations):
            for i in range(n):
                gradient= (1/m) * np.sum((X @ self.theta - y) * X[:,i].reshape(-1,1))
                self.theta[i]=self.theta[i]-self.alpha*gradient
      
        return self
        
    def predict(self,X):
        return (X @ self.theta).flatten()
```

# Using Single Loop

# J(w)=(1/m) * X.T * (Xw-y)

```python
class LinearRegressionGD:
    def __init__(self,alpha=0.01,iterations=100):
        self.alpha=alpha,
        self.iterations=iterations,
        self.theta=None

    def fit(self,X,y):
        m,n=X.shape
        y=y.reshape(-1,1)
        self.theta=np.zeros((n,1))

        for _ in range(self.iterations):
            gradient=X.T @ (X @ self.theta - y)
            self.theta=self.theta - self.alpha*gradient
      
        return self
        
    def predict(self,X):
        return (X @ self.theta).flatten()
```
